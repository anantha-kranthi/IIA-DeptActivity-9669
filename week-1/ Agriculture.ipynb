```python
fruits = ["Apple", "Banana", "Cherry", "Mango"]

# Positive Indexing
print(fruits[0])   # Output: Apple
print(fruits[2])   # Output: Cherry

# Negative Indexing
print(fruits[-1])  # Output: Mango
print(fruits[-2])  # Output: Banana

# Slicing: [start:stop] (stop index is excluded)
print(fruits[1:3]) # Output: ['Banana', 'Cherry']
```

```python
crops = ["Paddy", "Wheat", "Maize"]

# Updating an existing element
crops[1] = "Cotton"
print(crops) # Output: ['Paddy', 'Cotton', 'Maize']
```

```python
numbers = [40, 10, 30, 20]

numbers.append(50)      # [40, 10, 30, 20, 50]
numbers.sort()          # [10, 20, 30, 40, 50]
numbers.pop()           # Removes 50 -> [10, 20, 30, 40]

print(numbers)          # Output: [10, 20, 30, 40]
```

```python
scores = [88, 92, 79, 95, 84]

print(len(scores))   # Total items      -> Output: 5
print(sum(scores))   # Sum of items     -> Output: 438
print(max(scores))   # Maximum value    -> Output: 95
print(min(scores))   # Minimum value    -> Output: 79

# Calculate Average
average = sum(scores) / len(scores)
print(average)       # Output: 87.6

```
